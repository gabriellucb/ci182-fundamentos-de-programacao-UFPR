###- Exercício 8

###- Exercício 9
Faça um programa que leia uma matriz 3x3 informada pelo usuário e verifique se ela é uma Matriz Identidade. Uma matriz é identidade se todos os elementos da diagonal principal são 1 e todos os outros elementos são 0. O programa deve imprimir "É Matriz Identidade" ou "Não é Matriz Identidade".

```python
# RESOLUÇÃO - EXERCÍCIO 9

matriz = []

for i in range(3):
    linha = []
    for j in range(3):
        valor = int(input(f"Valor [{i}][{j}]: "))
        linha.append(valor)
    matriz.append(linha)

eh_identidade = True

# Verificação
for i in range(3):
    for j in range(3):
        # Condição para os elementos da diagonal principal
        if i == j:
            if matriz[i][j] != 1:
                eh_identidade = False
        # Condição para todos os outros elementos
        else:
            if matriz[i][j] != 0:
                eh_identidade = False

if eh_identidade:
    print("É Matriz Identidade")
else:
    print("Não é Matriz Identidade")
```

###- Exercício 10
A transposta de uma matriz é obtida transformando as suas linhas em colunas. Escreva um programa que leia os elementos de uma matriz 2x3 informados pelo usuário. Em seguida, o programa deve construir uma nova matriz 3x2 que seja a transposta da primeira e imprimi-la na tela formatada.

```python
# RESOLUÇÃO - EXERCÍCIO 10

matriz_original = []

# Lendo uma matriz 2x3
for i in range(2):
    linha = []
    for j in range(3):
        valor = int(input(f"Valor [{i}][{j}]: "))
        linha.append(valor)
    matriz_original.append(linha)

matriz_transposta = []

# Na transposta, a nova matriz terá 3 linhas e 2 colunas.
# Por isso, invertemos a ordem dos laços para construir!
for j in range(3):
    nova_linha = []
    for i in range(2):
        # A nova linha pega as informações da coluna [j] antiga
        nova_linha.append(matriz_original[i][j])
    matriz_transposta.append(nova_linha)

print("\nMatriz Transposta:")
for linha in matriz_transposta:
    print(linha)
```

###- Exercício 12
A multiplicação de duas matrizes não é feita apenas multiplicando elemento por elemento correspondente. O elemento C[i][j] da matriz resultante é calculado multiplicando-se os elementos da linha i da primeira matriz pelos elementos da coluna j da segunda, e somando os resultados. Faça um programa que leia duas matrizes quadradas 2x2 e calcule a matriz resultante da multiplicação entre elas.

```python
# RESOLUÇÃO - EXERCÍCIO 12

print("Matriz 1:")
m1 = []
for i in range(2):
    linha = []
    for j in range(2):
        linha.append(int(input(f"Valor [{i}][{j}]: ")))
    m1.append(linha)

print("\nMatriz 2:")
m2 = []
for i in range(2):
    linha = []
    for j in range(2):
        linha.append(int(input(f"Valor [{i}][{j}]: ")))
    m2.append(linha)

# Inicia a matriz resultante vazia, preenchida com zeros
m_resultado = [[0, 0], [0, 0]]

# Algoritmo da multiplicação
for i in range(2):
    for j in range(2):
        m_resultado[i][j] = (m1[i][0] * m2[0][j]) + (m1[i][1] * m2[1][j])

print("\nMatriz Multiplicada:")
for linha in m_resultado:
    print(linha)
```

###- Exercício 13
Faça um programa que leia uma matriz 3x3 e calcule o seu determinante.

```python
# RESOLUÇÃO - EXERCÍCIO 13

matriz = []

for i in range(3):
    linha = []
    for j in range(3):
        linha.append(int(input(f"Digite o valor de [{i}, {j}]: ")))
    matriz.append(linha)

# Cálculo manual através da Regra de Sarrus para 3x3
diag_principal1 = matriz[0][0] * matriz[1][1] * matriz[2][2]
diag_principal2 = matriz[0][1] * matriz[1][2] * matriz[2][0]
diag_principal3 = matriz[0][2] * matriz[1][0] * matriz[2][1]

diag_secundaria1 = matriz[0][2] * matriz[1][1] * matriz[2][0]
diag_secundaria2 = matriz[0][0] * matriz[1][2] * matriz[2][1]
diag_secundaria3 = matriz[0][1] * matriz[1][0] * matriz[2][2]

soma_principais = diag_principal1 + diag_principal2 + diag_principal3
soma_secundarias = diag_secundaria1 + diag_secundaria2 + diag_secundaria3

determinante = soma_principais - soma_secundarias

print(f"Determinante: {determinante}")
```

###- Exercício 14
O conceito de inversão de matrizes é complexo, mas a verificação é simples. Duas matrizes quadradas A e B são consideradas inversas se, e somente se, o produto delas é exatamente a Matriz Identidade. Faça um programa que leia os valores reais de duas matrizes 2x2. Utilize a lógica da multiplicação do Exercício 12 para encontrar a matriz resultante e, em seguida, verifique se essa resultante é a Identidade. Imprima se as matrizes são ou não inversas.

```python
# RESOLUÇÃO - EXERCÍCIO 14

print("Matriz A:")
m_A = []
for i in range(2):
    linha = []
    for j in range(2):
        linha.append(float(input(f"Valor [{i}][{j}]: ")))
    m_A.append(linha)

print("\nMatriz B:")
m_B = []
for i in range(2):
    linha = []
    for j in range(2):
        linha.append(float(input(f"Valor [{i}][{j}]: ")))
    m_B.append(linha)

# Multiplica e guarda
m_mult = [[0.0, 0.0], [0.0, 0.0]]
for i in range(2):
    for j in range(2):
        # Função round(valor, 0) lida com as pequenas frações quebradas que o Python gera com floats
        m_mult[i][j] = round((m_A[i][0] * m_B[0][j]) + (m_A[i][1] * m_B[1][j]), 0)

# Verificação se o produto é a Matriz Identidade (Diagonal = 1, Resto = 0)
if m_mult[0][0] == 1.0 and m_mult[0][1] == 0.0 and m_mult[1][0] == 0.0 and m_mult[1][1] == 1.0:
    print("As matrizes são inversas!")
else:
    print("As matrizes NÃO são inversas!")
```

###- Desafio
Faça um programa que leia uma matriz 3x3 contendo os caracteres 'X', 'O' e '' (espaço vazio), representando o estado atual de um jogo da velha. O algoritmo deve analisar as linhas, as colunas e as duas diagonais do tabuleiro para sugerir a próxima jogada do jogador 'X'. A lógica da sugestão deve seguir uma prioridade:  Se o 'X' puder ganhar o jogo nesta rodada (tiver dois 'X' e um vazio na mesma reta), sugira essa posição.  Caso não possa ganhar agora, mas o 'O' estiver prestes a ganhar (dois 'O' e um vazio na mesma reta), sugira essa posição para bloquear a derrota.  Caso nenhuma das anteriores seja verdade, sugira qualquer posição vazia.

```python
# RESOLUÇÃO - DESAFIO

tabuleiro = []

print("Tabuleiro:")
for i in range(3):
    linha = []
    for j in range(3):
        # As entradas vazias neste exemplo foram dadas usando o texto "", converta na leitura
        valor = input(f"Valor [{i}][{j}]: ")
        if valor == '""':
            valor = ""
        linha.append(valor)
    tabuleiro.append(linha)

jogada_linha = -1
jogada_col = -1

# Prioridade 1: Tentar ganhar o jogo (Verifica onde tem dois 'X' e um 'vazio')
for i in range(3):
    # Checando Linhas
    if tabuleiro[i].count('X') == 2 and tabuleiro[i].count('') == 1:
        jogada_linha = i
        jogada_col = tabuleiro[i].index('')
    # Checando Colunas
    coluna_atual = [tabuleiro[0][i], tabuleiro[1][i], tabuleiro[2][i]]
    if coluna_atual.count('X') == 2 and coluna_atual.count('') == 1:
        jogada_linha = coluna_atual.index('')
        jogada_col = i

# Checando Diagonais para Vitória
diag_prin = [tabuleiro[0][0], tabuleiro[1][1], tabuleiro[2][2]]
if diag_prin.count('X') == 2 and diag_prin.count('') == 1:
    indice_vazio = diag_prin.index('')
    jogada_linha = indice_vazio
    jogada_col = indice_vazio

diag_sec = [tabuleiro[0][2], tabuleiro[1][1], tabuleiro[2][0]]
if diag_sec.count('X') == 2 and diag_sec.count('') == 1:
    indice_vazio = diag_sec.index('')
    jogada_linha = indice_vazio
    jogada_col = 2 - indice_vazio

# Prioridade 2: Bloquear (Apenas se a Prioridade 1 falhou, ou seja, se os índices continuam -1)
if jogada_linha == -1:
    for i in range(3):
        # Checando Linhas
        if tabuleiro[i].count('O') == 2 and tabuleiro[i].count('') == 1:
            jogada_linha = i
            jogada_col = tabuleiro[i].index('')
        # Checando Colunas
        coluna_atual = [tabuleiro[0][i], tabuleiro[1][i], tabuleiro[2][i]]
        if coluna_atual.count('O') == 2 and coluna_atual.count('') == 1:
            jogada_linha = coluna_atual.index('')
            jogada_col = i

    # Checando Diagonais para Bloqueio
    if diag_prin.count('O') == 2 and diag_prin.count('') == 1:
        indice_vazio = diag_prin.index('')
        jogada_linha = indice_vazio
        jogada_col = indice_vazio

    if diag_sec.count('O') == 2 and diag_sec.count('') == 1:
        indice_vazio = diag_sec.index('')
        jogada_linha = indice_vazio
        jogada_col = 2 - indice_vazio

# Prioridade 3: Posição vazia aleatória (Apenas se 1 e 2 falharam)
if jogada_linha == -1:
    for i in range(3):
        for j in range(3):
            if tabuleiro[i][j] == '':
                jogada_linha = i
                jogada_col = j

# Saída do Algoritmo
print(f"\nJogada sugerida: [{jogada_linha}][{jogada_col}]")
```
