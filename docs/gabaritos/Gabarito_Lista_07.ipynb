###- Exercício 1
Crie um programa que abra (ou crie) um arquivo chamado saudacao.txt em modo de escrita ("w"). Escreva a frase "Escrito no Python!" dentro dele e feche o arquivo. Lembre de escrever uma mensagem ao usuário quando criar o arquivo.

```python
# RESOLUÇÃO - EXERCÍCIO 1

# Abre o arquivo em modo de escrita ("w"). Se não existir, ele é criado.
arquivo = open("saudacao.txt", "w")

# Escreve a frase dentro do arquivo (boa prática colocar o \n)
arquivo.write("Escrito no Python!\n")

# Fechar o arquivo é obrigatório para salvar e liberar memória
arquivo.close()

# Mensagem de feedback ao usuário
print("Arquivo 'saudacao.txt' criado com sucesso!")
```

###- Exercício 2

###- Exercício 4
Crie um programa que utilize um laço para pedir ao usuário o nome de 3 convidados para uma festa. O programa deve abrir um arquivo chamado convidados.txt e gravar cada nome digitado em uma linha diferente do arquivo.

```python
# RESOLUÇÃO - EXERCÍCIO 4

# Abre para escrita
arquivo = open("convidados.txt", "w")

# Laço para ler 3 nomes
for i in range(3):
    nome = input(f"Nome do convidado {i+1}: ")
    # Grava o nome e pula linha para que o próximo não fique colado
    arquivo.write(nome + "\n")

arquivo.close()

print("Lista de convidados salva no arquivo 'convidados.txt'.")
```

###- Exercício 5
Crie um programa que leia o arquivo convidados.txt criado no exercício anterior usando o método .read().splitlines(). O programa deve imprimir a lista gerada e informar quantos convidados existem (usando a função len()).

###- Exercício 6
Crie um programa que abra arquivo chamado numeros.txt em modo de escrita ("w"). Utilize um laço for para gravar 5 números no arquivo, garantindo que cada número fique em uma linha diferente (use o \n). Abra o arquivo numeros.txt em modo de leitura. Utilize o método .read().splitlines() para extrair as linhas como uma lista. Converta cada elemento para inteiro, calcule a soma total de todos os números e imprima o resultado.

```python
# RESOLUÇÃO - EXERCÍCIO 6

# PARTE 1: Gravação
arquivo_gravacao = open("numeros.txt", "w")

for i in range(5):
    valor = input("Insira um valor: ")
    arquivo_gravacao.write(valor + "\n")

arquivo_gravacao.close()
print("Arquivo 'numeros.txt' gerado com sucesso.\n")

# PARTE 2: Leitura e Processamento
arquivo_leitura = open("numeros.txt", "r")
linhas = arquivo_leitura.read().splitlines()
arquivo_leitura.close()

soma = 0
for linha in linhas:
    # Todo texto de arquivo vem como String, é preciso converter para inteiro
    soma = soma + int(linha)

print(f"Soma dos números do arquivo é: {soma}")
```

###- Exercício 7
Sistemas comerciais armazenam dados organizados por delimitadores (como a vírgula). Faça um programa que peça ao usuário o Nome e a Idade de 3 pessoas. Salve esses dados em um arquivo chamado dados.csv no formato Nome, Idade em cada linha.

```python
# RESOLUÇÃO - EXERCÍCIO 7

arquivo = open("dados.csv", "w")

for i in range(1, 4):
    nome = input(f"Nome da pessoa {i}: ")
    idade = input("Idade: ")
    # O segredo do CSV: valores separados por um delimitador fixo (a vírgula)
    arquivo.write(f"{nome},{idade}\n")

arquivo.close()

print("Base de dados 'dados.csv' gerada com sucesso!")
```

###- Exercício 8
Crie um programa que abra o arquivo dados.csv em modo de leitura. O programa deve ler linha por linha, utilizar a função .split(",") para separar o nome da idade, e descobrir qual é o nome da pessoa mais velha da lista.

```python
# RESOLUÇÃO - EXERCÍCIO 8

arquivo = open("dados.csv", "r")
linhas_bd = arquivo.read().splitlines()
arquivo.close()

maior_idade = -1
nome_mais_velho = ""

for linha in linhas_bd:
    # A linha está assim: "Felipe,20". O split transforma em: ["Felipe", "20"]
    dados = linha.split(",")

    nome_atual = dados[0]
    idade_atual = int(dados[1]) # Converte a idade para calcular corretamente

    # Lógica de encontrar o maior
    if idade_atual > maior_idade:
        maior_idade = idade_atual
        nome_mais_velho = nome_atual

print(f"A pessoa mais velha do arquivo é {nome_mais_velho} com {maior_idade} anos.")
```

###- Exercício 9
Crie um programa que leia o arquivo numeros.txt gerado no exercício 6. O programa deve abrir um SEGUNDO arquivo chamado pares.txt em modo "w", e gravar nele apenas os números lidos do primeiro arquivo que sejam pares.

```python
# RESOLUÇÃO - EXERCÍCIO 9

# 1. Lê a base completa
arquivo_original = open("numeros.txt", "r")
linhas = arquivo_original.read().splitlines()
arquivo_original.close()

# 2. Prepara o arquivo de saída
arquivo_pares = open("pares.txt", "w")

for linha in linhas:
    numero = int(linha)

    # Se o resto da divisão por 2 for zero, é par
    if numero % 2 == 0:
        # Temos que converter de volta para string para gravar!
        arquivo_pares.write(str(numero) + "\n")

arquivo_pares.close()

print("Filtro concluído! Apenas os pares foram salvos em 'pares.txt'.")
```

###- Desafio
Em jogos, o inventário do personagem é frequentemente atualizado e consultado. Vamos simular isso utilizando os três modos de ficheiro ("w", "a", "r").

```python
# RESOLUÇÃO - DESAFIO

# 1. Iniciando com itens pré-definidos (Modo "w" sobrescreve tudo)
arquivo_inventario = open("mochila.txt", "w")
arquivo_inventario.write("Mapa\n")
arquivo_inventario.write("Moedas\n")
arquivo_inventario.close()

# 2. Laço de interação com o jogador
while True:
    escolha = input("Adicionar item (S/N)? ").upper()

    if escolha == 'S':
        novo_item = input("Qual o nome do item? ")

        # Abre em modo "a" para apenas anexar o item sem deletar o que já tem
        arquivo_adicao = open("mochila.txt", "a")
        arquivo_adicao.write(novo_item + "\n")
        arquivo_adicao.close()

        print("Item adicionado!\n")

    elif escolha == 'N':
        break # Encerra o laço

    else:
        print("Opção inválida.")

# 3. Exibição do resultado final (Modo "r")
arquivo_leitura = open("mochila.txt", "r")
lista_mochila = arquivo_leitura.read().splitlines()
arquivo_leitura.close()

print(f"\nInventário Atualizado: {lista_mochila}")
print(f"O inventário tem o total de {len(lista_mochila)} itens.")
```
