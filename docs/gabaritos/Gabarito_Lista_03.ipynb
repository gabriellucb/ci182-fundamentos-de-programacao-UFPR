###- Exercício 1

Você foi encarregado de programar o painel de lançamento de um foguete de testes. Usando um laço while, crie um programa que peça ao usuário um número inicial para a contagem regressiva. O programa deve imprimir o valor do contador, diminuir 1 a cada repetição e, quando o contador chegar a 0, imprimir a mensagem "FOGO!".

```python
# RESOLUÇÃO - EXERCÍCIO 1

# Lê o número inicial fornecido pelo usuário
contador = int(input("Insira o número inicial: "))

# O laço continua enquanto o contador for estritamente maior que 0
while contador > 0:
    print(contador)
    # Diminui 1 do contador a cada volta (passo fundamental para não gerar loop infinito)
    contador = contador - 1

# Quando o laço termina (contador chega a 0), imprime a mensagem final
print("FOGO!")
```

###- Exercício 2
Um pesquisador precisa somar o peso de 5 amostras de rocha coletadas no campo. Usando um laço while e uma variável acumuladora, crie um programa que peça o peso da amostra 5 vezes. Ao final do laço, imprima o peso total acumulado.

```python
# RESOLUÇÃO - EXERCÍCIO 2

# Variável contadora: controla quantas vezes o laço rodou
amostras_lidas = 0
# Variável acumuladora: guarda a soma dos pesos. DEVE ser iniciada antes do laço!
peso_total = 0.0

while amostras_lidas < 5:
    peso = float(input("Digite o peso da amostra (kg): "))

    # Acumula o peso digitado na caixa 'peso_total'
    peso_total = peso_total + peso

    # Incrementa o contador para registrar que mais uma amostra foi lida
    amostras_lidas = amostras_lidas + 1

print(f"Peso total acumulado: {peso_total} kg.")
```

###- Exercício 3
Em uma catraca de refeitório, o sistema não sabe quantos alunos vão entrar, mas sabe que a catraca trava quando o operador digita o código "0". Usando while, crie um programa que leia continuamente a matrícula do aluno (número inteiro). Pare de ler quando o número 0 for digitado. Ao final, use uma variável contadora para mostrar exatamente quantos alunos passaram (não conte o zero).

```python
# RESOLUÇÃO - EXERCÍCIO 3

total_alunos = 0

# Lemos a primeira matrícula antes de entrar no laço
matricula = int(input("Matrícula do aluno (0 para travar): "))

# O laço só continua se a matrícula for diferente de 0
while matricula != 0:
    # Se entrou no laço, é porque não é zero, então contamos o aluno
    total_alunos = total_alunos + 1

    # Lemos a próxima matrícula para a próxima verificação
    matricula = int(input("Matrícula do aluno (0 para travar): "))

print(f"Total de alunos registrados: {total_alunos}.")
```

O comando de repetição for, quando utilizado junto com a função range(), facilita muito a criação de laços onde sabemos exatamente o valor inicial, o valor final e o "passo" (o tamanho do salto entre um número e outro). Crie um programa que peça ao usuário um número inteiro limite. ###- Exercício 6
Escreva um programa para calcular a soma acumulada de uma sequência de números. O programa deve pedir ao usuário dois números inteiros: um valor inicial e um valor final (assuma que o valor inicial será sempre menor ou igual ao final). Usando um laço de repetição, o programa deve percorrer todos os números dentro desse intervalo fechado (ou seja, incluindo o número inicial e o número final) e somá-los passo a passo. Ao final da repetição, exiba o resultado total da soma.

```python
# RESOLUÇÃO - EXERCÍCIO 6

inicio = int(input("Digite o valor inicial: "))
fim = int(input("Digite o valor final: "))

soma_intervalo = 0

# O laço for vai do 'inicio' até o 'fim'. O + 1 é para incluir o último número.
for numero in range(inicio, fim + 1):
    soma_intervalo = soma_intervalo + numero

print(f"A soma do intervalo é: {soma_intervalo}")
```

###- Exercício 7
Escreva um programa que solicite ao usuário dois números inteiros, representando o início e o fim de um intervalo. Utilizando um laço de repetição, o programa deve analisar cada número dentro desse intervalo fechado (ou seja, incluindo os próprios números digitados). Durante a repetição, o programa deve fazer duas tarefas simultaneamente: somar os valores desses números divisíveis por 7 e registrar quantos foram encontrados. Ao final, imprima a soma total e a quantidade de números divisíveis por 7.

```python
# RESOLUÇÃO - EXERCÍCIO 7

inicio = int(input("Valor inicial: "))
fim = int(input("Valor final: "))

soma_divisiveis = 0
quantidade_divisiveis = 0

for numero in range(inicio, fim + 1):
    # Verifica se o resto da divisão por 7 é zero
    if numero % 7 == 0:
        soma_divisiveis = soma_divisiveis + numero
        quantidade_divisiveis = quantidade_divisiveis + 1

print(f"Soma dos divisíveis por 7: {soma_divisiveis}")
print(f"Quantidade de números encontrados: {quantidade_divisiveis}")
```

###- Exercício 8
Você foi encarregado de desenvolver um software para auxiliar um professor a calcular a média das notas de uma turma. Como o professor não sabe de antemão quantos alunos existem na sala, o programa deve pedir as notas continuamente. A repetição só deve ser interrompida quando o professor digitar um valor negativo (por exemplo, -1), que servirá como um "sinal de parada". O programa deve acumular a soma das notas válidas e contar quantas notas foram digitadas. Ao final, calcule e imprima a média aritmética da turma.

```python
# RESOLUÇÃO - EXERCÍCIO 8

soma_notas = 0.0
quantidade_alunos = 0

nota = float(input("Informe a nota do aluno (-1 para encerrar): "))

# O laço roda enquanto a nota for válida (maior ou igual a zero)
while nota >= 0:
    soma_notas = soma_notas + nota
    quantidade_alunos = quantidade_alunos + 1

    # Pede a próxima nota no final do laço
    nota = float(input("Informe a nota do aluno (-1 para encerrar): "))

# Proteção para não dividir por zero caso o professor digite -1 logo de cara
if quantidade_alunos > 0:
    media = soma_notas / quantidade_alunos
    print(f"Média da turma: {media:.1f}")
else:
    print("Nenhuma nota válida foi digitada.")
```

###- Exercício 9
Crie um programa que simule um caixa de mercado usando while True. O sistema deve pedir o valor de cada produto comprado. Se o valor digitado for 0, o laço deve ser encerrado. Durante o processo, acumule o valor total da compra. Quando o laço acabar, se o total da compra for maior que R$ 200.00, aplique um desconto de 10%. Imprima o valor final a ser pago.

```python
# RESOLUÇÃO - EXERCÍCIO 9

total_compra = 0.0

# O while True cria um laço infinito que só para quando encontra o 'break'
while True:
    valor_produto = float(input("Valor do produto (0 para finalizar): "))

    # Condição de parada (break)
    if valor_produto == 0:
        break

    total_compra = total_compra + valor_produto

# Fora do laço, verificamos se há desconto
if total_compra > 200.00:
    # Aplica 10% de desconto (multiplicar por 0.9 é o mesmo que tirar 10%)
    total_compra = total_compra * 0.90

print(f"Total a pagar com desconto: R$ {total_compra:.2f}")
```

###- Exercício 10
Crie um sistema de segurança utilizando o laço while True (que cria uma repetição infinita por padrão). O programa deve pedir a senha ao usuário, mas ele possui um limite estrito de 5 tentativas. Dentro do laço, o seu código deve seguir estas regras: Se a senha digitada for "UFPR2026", imprima "Acesso Liberado" e encerre o programa na hora. Caso a senha esteja errada, reduza o número de tentativas e avise o usuário de quantas chances ele ainda tem. Se as tentativas acabarem, o programa também deve ser encerrado.

```python
# RESOLUÇÃO - EXERCÍCIO 10

tentativas = 5

while True:
    senha = input("Digite a senha de acesso: ")

    if senha == "UFPR2026":
        print("Acesso Liberado")
        break # Quebra o laço porque a senha está correta
    else:
        tentativas = tentativas - 1

        # Verifica se ainda tem tentativas antes de perguntar de novo
        if tentativas > 0:
            print(f"Senha incorreta, restam {tentativas} tentativas.")
        else:
            print("Senha incorreta. Acesso bloqueado!")
            break # Quebra o laço porque acabaram as chances
```

###- Exercício 11
O fatorial de um número inteiro positivo (representado por N!) é a multiplicação desse número por todos os seus antecessores até o número 1. Escreva um programa que receba um número e, utilizando um laço de repetição (while ou for), calcule o seu Fatorial (produto acumulado).

```python
# RESOLUÇÃO - EXERCÍCIO 11

numero = int(input("Digite um número para calcular o fatorial: "))

# A variável de produto acumulado NÃO pode começar em 0 (pois 0 * x = 0).
fatorial = 1

# Usamos um for de 1 até o próprio número
for i in range(1, numero + 1):
    fatorial = fatorial * i

print(f"O fatorial é {fatorial}")
```

###- Exercício 12
Sistemas industriais precisam de dados confiáveis. Escreva um programa que obrigue o operador a preencher duas informações de forma correta usando laços while:
1. Temperatura de Operação: Não pode ser menor que 0 nem maior que 1000.
2. Pressão da Válvula: Deve ser um valor estritamente maior que 0.

Caso o operador digite um valor inválido, o programa deve pedir o mesmo dado novamente até que seja válido. Só imprima "Configuração aceita" quando ambos estiverem corretos.

```python
# RESOLUÇÃO - EXERCÍCIO 12

# Lê a temperatura a primeira vez
temperatura = float(input("Temperatura de Operação (0 a 1000): "))

# Enquanto a temperatura for menor que 0 OU maior que 1000, repete a pergunta
while temperatura < 0 or temperatura > 1000:
    print("Erro! Temperatura inválida.")
    temperatura = float(input("Temperatura de Operação (0 a 1000): "))

# Lê a pressão a primeira vez
pressao = float(input("Pressão da Válvula (>0): "))

# Enquanto a pressão for menor ou igual a 0, repete a pergunta
while pressao <= 0:
    print("Erro! Pressão inválida.")
    pressao = float(input("Pressão da Válvula (>0): "))

# Só chega nesta linha se passar ileso pelos dois whiles de validação
print("Configuração aceita.")
```

###- Exercício 13
Em uma linha de produção industrial, o controle de qualidade avalia os lotes analisando 10 peças por vez. Crie um programa que utilize o laço de repetição for para solicitar ao operador o comprimento (em centímetros) de exatamente 10 peças. Durante a repetição, o programa deve classificar e contar cada peça com base nas seguintes regras:

1. Tamanho Ideal: O comprimento deve estar entre 10.0 cm e 10.5 cm (incluindo esses valores).
2. Defeituosa: Qualquer comprimento que esteja fora desse intervalo.

Ao final do laço (após a leitura e classificação de todas as 10 peças), o programa deve exibir: a quantidade de peças com tamanho ideal, a quantidade de peças defeituosas e a porcentagem de peças defeituosas do lote.

```python
# RESOLUÇÃO - EXERCÍCIO 13

pecas_ideais = 0
pecas_defeituosas = 0

# O laço repete exatamente 10 vezes
for i in range(10):
    comprimento = float(input("Comprimento da peça: "))

    # Classificação
    if 10.0 <= comprimento <= 10.5:
        pecas_ideais = pecas_ideais + 1
    else:
        pecas_defeituosas = pecas_defeituosas + 1

# Cálculo da porcentagem (cada peça vale 10% do lote de 10)
porcentagem_perda = (pecas_defeituosas / 10) * 100

print(f"Peças ideais: {pecas_ideais}")
print(f"Peças defeituosas: {pecas_defeituosas}")
print(f"{porcentagem_perda:.1f}% de perda.")
```

###- Exercício 14 (Desafio)
Em sistemas de criptografia, a geração de números primos é uma etapa fundamental. Um número primo é aquele que só é divisível exatamente por 1 e por ele mesmo. Escreva um programa que leia um número inteiro "limite" maior que 1. O programa deve analisar e imprimir todos os números primos que existem entre 2 e o número limite digitado.

```python
# RESOLUÇÃO - EXERCÍCIO 14 (DESAFIO)

limite = int(input("Digite o número limite: "))

print(f"Os números primos de 2 até {limite} são:")

# Laço Externo: Seleciona um número de 2 até o limite estipulado
for numero_atual in range(2, limite + 1):

    # O contador zera A CADA NÚMERO NOVO que vamos testar
    divisores = 0

    # Laço Interno: Tenta dividir o número atual por todos os números de 1 até ele mesmo
    for divisor in range(1, numero_atual + 1):
        if numero_atual % divisor == 0:
            divisores = divisores + 1

    # Se ao final dos testes, ele teve exatamente 2 divisores (1 e ele mesmo), é Primo!
    if divisores == 2:
        print(numero_atual)
```
