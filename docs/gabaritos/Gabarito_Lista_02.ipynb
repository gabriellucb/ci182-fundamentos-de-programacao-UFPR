### Exercício 2

Crie um programa de análise matemática que receba um número inteiro qualquer fornecido pelo usuário. O programa deve informar se ele pertence à categoria dos números **Pares** ou dos números **Ímpares** e também verificar se o número lido é o **zero**. Além disso, o programa deve imprimir na tela se o número é **positivo** ou **negativo**.

```python
# RESOLUÇÃO - EXERCÍCIO 2

numero = int(input("Digite um número inteiro: "))

# Verificação de Par ou Ímpar
if numero == 0:
    print(f"O número {numero} é zero (par).")
elif numero % 2 == 0:
    print(f"O número {numero} é Par.")
else:
    print(f"O número {numero} é Ímpar.")

# Verificação de Positivo ou Negativo
if numero > 0:
    print("O número é positivo.")
elif numero < 0:
    print("O número é negativo.")
```

### Exercício 3

Um sistema de uma indústria química classifica os reagentes pelo pH:
* pH < 7: "Ácido"
* pH == 7: "Neutro"
* pH > 7: "Básico"

O programa deve ler o pH e garantir que o valor esteja entre 0 e 14. Se estiver fora disso, imprima "Erro: pH inválido".


```python
# RESOLUÇÃO - EXERCÍCIO 3

ph = float(input("pH: "))

# Validação do intervalo permitido (0 a 14)
if ph < 0 or ph > 14:
    print("Erro: pH inválido")
else:
    # Classificação baseada no valor do pH
    if ph < 7:
        print("Ácido")
    elif ph == 7:
        print("Neutro")
    else:
        print("Básico")
```

### Exercício 5

Uma empresa de logística dá desconto baseado na distância. Faça um programa que peça a distância em quilômetros e calcule o valor seguindo esses critérios:
* **Até 100km:** R$ 50,00 fixo.

* **Entre 100km e 300km:** R$ 0,45 por km.

* **Acima de 300km:** R$ 0,35 por km.


```python
# RESOLUÇÃO - EXERCÍCIO 5

distancia = float(input("Insira a distância (em quilômetros): "))

if distancia <= 100:
    valor_frete = 50.00
elif distancia <= 300:
    valor_frete = distancia * 0.45
else:
    valor_frete = distancia * 0.35

# O :.2f Faz com que existam apenas duas casas decimais.
print(f"Valor do frete: R$ {valor_frete:.2f}")
```

Desenvolva um algoritmo que solicite ao usuário um ano e determine se ele é Bissexto.
* O ano deve ser divisível por 4;
* PORÉM, ele não pode ser divisível por 100;
* A MENOS QUE ele também seja divisível por 400.

**Dica:** Utilize os operadores lógicos AND e OR. Use parênteses para definir a precedência.

### Exercício 9

Você está desenvolvendo o software de monitoramento para uma estação de tratamento de água. O sistema lê continuamente o nível de Cloro Residual (mg/L) e verifica o estado de um Botão de Interrupção de Emergência.

O programa deve solicitar dois valores:
1. O nível de Cloro (número decimal).
2. O estado da interrupção: digite **1** para Ativado ou **0** para Desativado.

**Regras de Operação:**
* Se a interrupção estiver ‘Ativada’ (1), o sistema deve ignorar a leitura do cloro e exibir: "Operação Suspensa: Sistema paralisado pelo operador".
* Se a interrupção NÃO estiver ativa (0), o programa deve avaliar o cloro:
    * **Entre 0.2 e 2.0 (inclusive):** Exibir "Monitoramento: Nível Ideal".
    * **Fora desse intervalo:** Exibir "Alerta: Nível de cloro fora dos padrões. Verifique o dosador".


```python
# RESOLUÇÃO - EXERCÍCIO 9

cloro = float(input("Nível de Cloro (mg/L): "))
interrupcao_input = int(input("Interrupção (1 = Ativado / 0 = Desativado): "))

# Convertendo o número 1 ou 0 para Booleano (True ou False)
# Em Python, bool(1) vira True e bool(0) vira False
interrupcao = bool(interrupcao_input)

if interrupcao:
    # Se for True (1)
    print("Operação Suspensa: Sistema paralisado pelo operador")
else:
    # Se for False (0), avaliamos o cloro
    if 0.2 <= cloro <= 2.0:
        print("Monitoramento: Nível Ideal")
    else:
        print("Alerta: Nível de cloro fora dos padrões. Verifique o dosador")
```

### Exercício 10 (Desafio)

Implemente as regras de decisão para um elevador industrial:
1. **Prioridade Máxima (Emergência):** Se ativo, o elevador trava.
2. **Segurança de Cabine:** Porta aberta com peso impede o movimento.
3. **Sobrecarga Crítica:** Peso > 1000kg impede o movimento.
4. **Contradição de Sensor:** Peso <= 0 com porta fechada indica falha do sensor.

Exiba o STATUS respeitando a hierarquia de prioridades.

```python
# RESOLUÇÃO - EXERCÍCIO 10

peso = int(input("Peso da carga (kg): "))

# Lemos como inteiro (0 ou 1) e convertemos para Booleano
# bool(1) vira True | bool(0) vira False
porta_input = int(input("Porta aberta? (1-Sim / 0-Não): "))
porta = bool(porta_input)

emergencia_input = int(input("Botão Emergência ativo? (1-Sim / 0-Não): "))
emergencia = bool(emergencia_input)

# Hierarquia de decisão (A ordem aqui define a prioridade do sistema)
if emergencia:
    # Prioridade Máxima: Se houver emergência, nada mais importa.
    print("STATUS: BLOQUEIO TOTAL")

elif porta:
    # Só chega aqui se 'emergencia' for False.
    # Se a porta estiver aberta (True), o elevador não move.
    print("STATUS: ERRO DE PORTA")

elif peso > 1000:
    # Só chega aqui se 'emergencia' e 'porta' estiverem seguras (False).
    print("STATUS: SOBRECARGA")

elif peso <= 0:
    # Se o peso for zero ou negativo COM A PORTA FECHADA (pois o elif acima já testou porta aberta).
    print("STATUS: SENSOR DE PESO AVARIADO")

else:
    # Condição ideal: Sem emergência, porta fechada e peso dentro do limite.
    print("STATUS: LIBERADO PARA SUBIR")
```
