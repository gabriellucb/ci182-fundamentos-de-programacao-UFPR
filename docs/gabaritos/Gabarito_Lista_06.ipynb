###- Exercício 3
Crie uma função que receba um número inteiro positivo, calcule e retorne o seu fatorial usando um laço de repetição. Lembre-se que o fatorial de 0 é 1.

```python
# RESOLUÇÃO - EXERCÍCIO 3

def fatorial(n):
    # Trata o caso especial do 0 e do 1
    if n == 0 or n == 1:
        return 1

    resultado = 1
    # O laço vai de 1 até n (precisamos colocar n + 1 no range)
    for i in range(1, n + 1):
        resultado = resultado * i

    return resultado

# Programa Principal
num = int(input("Digite um número para calcular o fatorial: "))
print(f"O fatorial de {num} é {fatorial(num)}")
```

###- Exercício 4
Crie uma função que receba uma lista contendo vários números inteiros. A função deve retornar uma nova lista contendo apenas os números pares da lista original.

```python
# RESOLUÇÃO - EXERCÍCIO 4

def filtra_pares(lista_numeros):
    lista_pares = []
    for num in lista_numeros:
        # Se o resto da divisão por 2 for 0, o número é par
        if num % 2 == 0:
            lista_pares.append(num)
    return lista_pares

# Programa Principal
numeros_originais = [1, 2, 3, 4, 5, 6]
apenas_pares = filtra_pares(numeros_originais)

print(f"Lista original: {numeros_originais}")
print(f"Lista filtrada (apenas pares): {apenas_pares}")
```

###- Exercício 5
Crie uma função que recebe uma lista de números inteiros e retorna, de uma só vez, os dois valores: o maior e o menor número da lista. No programa principal, capture esses dois retornos em duas variáveis separadas e exiba.

```python
# RESOLUÇÃO - EXERCÍCIO 5

def maior_e_menor(lista):
    # Usando as funções nativas do Python para simplificar
    maior = max(lista)
    menor = min(lista)
    # Python permite retornar múltiplos valores separados por vírgula
    return maior, menor

# Programa Principal
valores = [45, 12, 89, 2, 34]

# Capturando os retornos na exata ordem que foram enviados
valor_maximo, valor_minimo = maior_e_menor(valores)

print(f"O maior é {valor_maximo} e o menor é {valor_minimo}")
```

###- Exercício 6
Construa uma calculadora utilizando o conceito de modularização. Crie quatro funções matemáticas: [1] soma(a, b), [2] subtrai(a, b), [3] multiplica(a, b) e [4] divide(a, b). No programa principal, crie um laço com um Menu interativo para o usuário escolher a operação. De acordo com a escolha, peça os números, chame a função correta e exiba o resultado. Não se esqueça de tratar a divisão por zero.

```python
# RESOLUÇÃO - EXERCÍCIO 6

# Definindo as 4 funções básicas
def soma(a, b):
    return a + b
def subtrai(a, b):
    return a - b
def multiplica(a, b):
    return a * b
def divide(a, b):
    if b == 0:
        return "Erro: Divisão por zero!"
    return a / b

# Programa Principal (Apenas um exemplo)
while True:
    print("\n--- MENU ---")
    print("[1] Somar")
    print("[2] Subtrair")
    print("[3] Multiplicar")
    print("[4] Dividir")
    print("[0] Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == '0':
        print("Encerrando calculadora...")
        break

    if opcao in ['1', '2', '3', '4']:
        num1 = float(input("Primeiro número: "))
        num2 = float(input("Segundo número: "))

        if opcao == '1': print(f"Resultado: {soma(num1, num2)}")
        elif opcao == '2': print(f"Resultado: {subtrai(num1, num2)}")
        elif opcao == '3': print(f"Resultado: {multiplica(num1, num2)}")
        elif opcao == '4': print(f"Resultado: {divide(num1, num2)}")
    else:
        print("Opção inválida!")
```

###- Exercício 7
Crie uma função valida_senha(senha) que recebe uma string. A função deve retornar True se a senha for forte, e False caso contrário. Regras para senha forte:

Ter no mínimo 8 caracteres.

Não conter espaços em branco.

Conter pelo menos um caractere especial: "@", "!", ou "#".
Utilize a nova função no programa principal.

```python
# RESOLUÇÃO - EXERCÍCIO 7

def valida_senha(senha):
    # Regra 1: Tamanho mínimo
    if len(senha) < 8:
        return False

    # Regra 2: Sem espaços
    if " " in senha:
        return False

    # Regra 3: Conter pelo menos um dos especiais
    if "@" not in senha and "!" not in senha and "#" not in senha:
        return False

    # Se não caiu em nenhum 'return False' acima, a senha é forte
    return True

# Programa Principal
senha_teste = input("Digite a senha para validação: ")

if valida_senha(senha_teste):
    print("Aprovado: Senha Forte!")
else:
    print("Recusado: A senha não atende aos requisitos de segurança.")
```

###- Exercício 8
Crie duas funções: Uma que cria uma matriz com os valores que o usuário digitar e uma que recebe essa matriz (lista de listas) e a imprime na tela formatada linha por linha (não apenas usando o print bruto da lista).

```python
# RESOLUÇÃO - EXERCÍCIO 8

def cria_matriz(linhas, colunas):
    matriz = []
    for i in range(linhas):
        linha_atual = []
        for j in range(colunas):
            valor = int(input(f"Valor [{i}][{j}]: "))
            linha_atual.append(valor)
        matriz.append(linha_atual)
    return matriz

def imprime_matriz(matriz):
    for linha in matriz:
        linha_formatada = ""
        for elemento in linha:
            # Formata garantindo espaçamento entre as 'paredes'
            linha_formatada += f"| {elemento} "
        print(linha_formatada + "|")

# Programa Principal
print("=== Gerador de Matriz ===")
minha_matriz = cria_matriz(3, 3)

print("\nMatriz Gerada:")
imprime_matriz(minha_matriz)
```

###- Desafio 1
A lógica por trás dos números romanos exige olhar não apenas para a letra atual, mas para a próxima. Crie uma função romano_para_inteiro(romano) que receba uma string representando um número romano e retorne o seu valor inteiro. As letras têm valores fixos (I=1, V=5, X=10, L=50, C=100, D=500, M=1000). Se uma letra de menor valor vem antes de uma de maior valor, ela deve ser subtraída (Ex: "IV" = 4). Caso contrário, é somada (Ex: "VI" = 6). Utilize quantas funções auxiliares forem necessárias.

```python
# RESOLUÇÃO - DESAFIO 1

def valor_letra(letra):
    # Função auxiliar que devolve o valor numérico puro
    letra = letra.upper()
    if letra == 'I': return 1
    if letra == 'V': return 5
    if letra == 'X': return 10
    if letra == 'L': return 50
    if letra == 'C': return 100
    if letra == 'D': return 500
    if letra == 'M': return 1000
    return 0

def romano_para_inteiro(romano):
    total = 0
    tamanho = len(romano)

    for i in range(tamanho):
        valor_atual = valor_letra(romano[i])

        # O segredo: Se existir uma letra à direita e ela for maior que a atual, subtraímos
        if i + 1 < tamanho and valor_letra(romano[i + 1]) > valor_atual:
            total = total - valor_atual
        else:
            total = total + valor_atual

    return total

# Programa Principal
numero_romano = input("Digite um número romano: ")
resultado = romano_para_inteiro(numero_romano)

print(f"O valor inteiro de {numero_romano.upper()} é {resultado}")
```

###- Desafio 2
Em qualquer sistema comercial ou banco de dados no Brasil, validar um CPF é uma etapa obrigatória. Um CPF não é apenas validado pelo seu tamanho (11 dígitos), mas por um cálculo matemático rigoroso que gera os dois últimos "Dígitos Verificadores". Crie um programa utilizando modularização contendo duas funções:

limpa_cpf(cpf_sujo): Recebe a string que o usuário digitou e devolve apenas os números, removendo pontos e traços.

valida_cpf(cpf): Implementa o algoritmo oficial de validação.

```python
# RESOLUÇÃO - DESAFIO 2

def limpa_cpf(cpf_sujo):
    cpf_limpo = ""
    for caractere in cpf_sujo:
        # Pega apenas o que for número, ignorando pontos e traços
        if caractere.isdigit():
            cpf_limpo += caractere
    return cpf_limpo

def valida_cpf(cpf_string):
    # Limpa a string antes de qualquer conta
    cpf = limpa_cpf(cpf_string)

    # Validações básicas de tamanho e números repetidos (111.111...)
    if len(cpf) != 11 or cpf == cpf[0] * 11:
        return False

    # Cálculo do PRIMEIRO dígito verificador
    soma = 0
    peso = 10
    for i in range(9):
        soma += int(cpf[i]) * peso
        peso -= 1

    digito_1 = (soma * 10) % 11
    if digito_1 == 10:
        digito_1 = 0

    if digito_1 != int(cpf[9]):
        return False

    # Cálculo do SEGUNDO dígito verificador
    soma = 0
    peso = 11
    for i in range(10):
        soma += int(cpf[i]) * peso
        peso -= 1

    digito_2 = (soma * 10) % 11
    if digito_2 == 10:
        digito_2 = 0

    if digito_2 != int(cpf[10]):
        return False

    return True

# Programa Principal
cpf_teste = input("Digite um CPF (com ou sem pontuação): ")

if valida_cpf(cpf_teste):
    print("Status: CPF VÁLIDO.")
else:
    print("Status: CPF INVÁLIDO.")
```
