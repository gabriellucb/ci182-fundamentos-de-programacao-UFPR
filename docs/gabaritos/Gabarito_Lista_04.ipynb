###- Exercício 1

Um laboratório de meteorologia registra a temperatura de uma cidade ao longo de 7 dias na lista temperaturas = [22.5, 23.0, 21.5, 24.0, 25.5, 20.0, 19.5]. Escreva um programa que acesse e imprima exclusivamente a temperatura do primeiro e do último dia registrado, de forma dinâmica (mesmo se o tamanho da lista mudar).

###- Exercício 5

Após uma competição, os juízes lançaram as pontuações: pontos = [120, 85, 300, 210, 95]. Organize a lista da menor para a maior nota e extraia o valor máximo (grande vencedor) imprimindo ambos os resultados.

```python
# RESOLUÇÃO - EXERCÍCIO 5

# Lista de pontuações desordenada
pontos = [120, 85, 300, 210, 95]

# A função .sort() organiza a própria lista do menor para o maior (crescente)
pontos.sort()

# Como está em ordem crescente, o maior valor está garantido na última posição (índice -1)
vencedor = pontos[-1]

# Exibe a classificação e o vencedor
print(f"Classificação geral: {pontos}")
print(f"Pontuação do grande vencedor: {vencedor}")
```

###- Exercício 6

Crie um programa que inicie com uma lista vazia e, utilizando um laço de repetição, peça para o usuário digitar o nome de produtos continuamente e os adicione ao final da lista. A leitura deve parar quando o usuário quiser. Ao término do laço, exiba o carrinho.

```python
# RESOLUÇÃO - EXERCÍCIO 6

# Inicia uma lista vazia para armazenar o carrinho
carrinho = []

# Laço indeterminado para pedir produtos continuamente
while True:
    produto = input("Digite o nome do produto (ou 'FIM' para parar): ")

    # Condição de parada do laço
    if produto.upper() == 'FIM':
        break

    # O .append() empurra o novo produto para a última posição da lista
    carrinho.append(produto)

# Exibe o carrinho no final do processo
print(f"Carrinho de compras: {carrinho}")
```

###- Exercício 7

Um sistema precisa ler o preço de vários produtos que um cliente está comprando. Crie um programa que inicie uma lista vazia e utilize um laço indeterminado para pedir valores. O sistema deve continuar armazenando os valores na lista até que o operador digite 0. Mostre a lista de preços e a soma deles no final.

```python
# RESOLUÇÃO - EXERCÍCIO 7

# Inicia a lista vazia para armazenar os preços
precos = []

# Laço infinito, o programa fica lendo valores até a condição de parada
while True:
    valor = float(input("Digite o preço do produto (0 para finalizar): "))

    # Condição de parada: se o usuário digitar 0, quebramos o laço com break
    if valor == 0:
        break

    # Se não for 0, o valor é armazenado na lista
    precos.append(valor)

# A função nativa sum() soma automaticamente todos os números de uma lista
# Pode ser feito sem uso da função sum()
valor_final = sum(precos)

# Exibe a lista final e a soma formatada com duas casas decimais
print(f"Produtos registrados: {precos}")
print(f"Valor final: R$ {valor_final:.2f}")
```

###- Exercício 8

Crie um programa que inicie uma lista para notas de um aluno. Utilize um laço para solicitar valores positivos continuamente (digitar valores negativos deve parar a leitura). Ao final, calcule a média geral e exiba as notas inseridas e a média final.

```python
# RESOLUÇÃO - EXERCÍCIO 8

# Inicia a lista de notas vazia
notas = []

# Laço para ler valores continuamente
while True:
    nota = float(input("Digite uma nota (ou negativo para encerrar): "))

    # Condição de parada: qualquer valor menor que zero
    if nota < 0:
        break

    notas.append(nota)

# Verifica se a lista não está vazia (para evitar erro matemático de divisão por zero)
if len(notas) > 0:
    # Média = soma de todos os itens / quantidade de itens na lista
    media = sum(notas) / len(notas)

    print(f"Notas registradas: {notas}")
    print(f"A média geral do aluno é: {media}")
else:
    print("Nenhuma nota válida foi inserida.")
```

###- Exercício 9

Crie um programa que leia vários valores numéricos de ações informados pelo usuário e os armazene em uma lista. A leitura deve prosseguir até o usuário digitar o valor 0. Após o fim da leitura encontre e imprima o maior e o menor valor lido na lista.

```python
# RESOLUÇÃO - EXERCÍCIO 9

# Inicia a lista vazia para os preços das ações
precos_acoes = []

# Laço de leitura contínua
while True:
    preco = float(input("Digite o preço da ação (ou 0 para sair): "))

    # Condição de parada
    if preco == 0:
        break

    precos_acoes.append(preco)

# Após o laço, verifica se há dados para fazer a análise
if len(precos_acoes) > 0:
    # Funções nativas max() e min() encontram o maior e menor valor de uma lista
    maior_preco = max(precos_acoes)
    menor_preco = min(precos_acoes)

    print(f"Preços registrados: {precos_acoes}")
    print(f"O maior preço registrado foi: {maior_preco}")
    print(f"O menor preço registrado foi: {menor_preco}")
else:
    print("Nenhum preço foi registrado.")
```

###- Exercício 10

Peça ao usuário diversas idades em um laço de repetição contínuo e armazene-as em uma lista (pare a leitura quando for digitado um valor negativo). Para fins legais, seu programa precisa distribuir essas idades em duas novas listas: uma apenas para maiores de idade e outra para menores de idade. No final, imprima as duas novas listas.

```python
# RESOLUÇÃO - EXERCÍCIO 10

# Inicia as duas listas vazias que receberão a distribuição
maiores = []
menores = []

# Laço de repetição contínuo para leitura
while True:
    idade = int(input("Digite a idade (ou um valor negativo para sair): "))

    # Condição de parada
    if idade < 0:
        break

    # Estrutura condicional para decidir para qual lista a idade vai
    if idade >= 18:
        maiores.append(idade)
    else:
        menores.append(idade)

# Exibe as estruturas consolidadas
print(f"Maiores de idade: {maiores}")
print(f"Menores de idade: {menores}")
```

###- Exercício 11

Uma fábrica separa peças: identificadores pares vão para o "Lote A" e ímpares para o "Lote B". Faça um programa que peça ao usuário diversos números inteiros positivos (parando ao digitar um negativo). O programa deve separar os dados em duas listas de lotes distintos e exibi-las no final.

```python
# RESOLUÇÃO - EXERCÍCIO 11

# Listas correspondentes a cada esteira/lote
lote_a = [] # Pares
lote_b = [] # Ímpares

while True:
    numero = int(input("Digite um número de série (negativo para parar): "))

    # Condição de parada com número negativo
    if numero < 0:
        break

    # Operador módulo (%) analisa o resto da divisão. Se for 0, é par.
    if numero % 2 == 0:
        lote_a.append(numero)
    else:
        lote_b.append(numero)

print(f"Lote A (Pares): {lote_a}")
print(f"Lote B (Ímpares): {lote_b}")
```

###- Exercício 12

Crie duas listas paralelas: produtos = ["Maçã", "Banana", "Leite"] e precos = [3.50, 2.00, 5.00]. O programa deve solicitar o nome de um produto. Se ele existir na primeira lista, localize em qual índice ele está e use esse índice para resgatar e imprimir o seu valor na segunda lista.

```python
# RESOLUÇÃO - EXERCÍCIO 12

# Definição das listas paralelas (os índices de um produto e seu preço são idênticos)
produtos = ["Maçã", "Banana", "Leite"]
precos = [3.50, 2.00, 5.00]

# Pede a string de busca para o usuário
busca = input("Digite o produto que deseja consultar: ")

# Checa se o produto consta no sistema (na lista de nomes)
if busca in produtos:
    # A função .index() retorna o número da posição (índice) onde aquele item está
    indice = produtos.index(busca)

    # Usamos o índice descoberto para 'pescar' o valor correspondente na lista de preços
    valor = precos[indice]

    print(f"O valor do produto {busca} é R$ {valor:.2f}")
else:
    print("Produto não cadastrado.")
```

###- Exercício 13

Faça um programa que construa uma matriz 3x3 utilizando listas aninhadas. Utilize laços de repetição (for) para varrer as linhas e colunas (uma matriz 3x3 tem posições fixas, justificando o for), pedindo para o usuário informar o valor de cada posição. Ao final, percorra a matriz construída e imprima-a formatada na tela, linha por linha.

```python
# RESOLUÇÃO - EXERCÍCIO 13

# Inicia a matriz principal (que guardará as listas menores)
matriz = []

# O laço externo controla as 3 linhas da nossa matriz
for linha in range(3):
    nova_linha = [] # Para cada nova iteração, cria uma linha vazia

    # O laço interno controla as 3 colunas da linha atual
    for coluna in range(3):
        valor = int(input(f"Digite o valor de [{linha}, {coluna}]: "))
        # Adiciona o elemento individual na linha
        nova_linha.append(valor)

    # Ao terminar a linha, anexa a lista inteira dentro da matriz principal
    matriz.append(nova_linha)

print("\nMatriz 3x3 gerada:")

# Percorre a lista principal, imprimindo cada sublista inteira por linha
for linha in matriz:
    print(linha)
```

###- Exercício 14

Um laboratório precisa identificar se uma sequência de DNA é um palíndromo (se a lista lida da esquerda para a direita é igual à lida da direita para a esquerda). Crie um programa que peça para o usuário inserir letras continuamente em uma lista (parando ao digitar "0") e faça essa verificação.

```python
# RESOLUÇÃO - EXERCÍCIO 14

# Lista para armazenar as letras do fragmento de DNA
fragmento = []

# Laço contínuo para solicitar letras de tamanho variável
while True:
    letra = input("Insira o Fragmento (ou '0' para finalizar): ")

    # Condição de parada
    if letra == '0':
        break

    fragmento.append(letra)

# Verifica se há dados para analisar
if len(fragmento) > 0:
    # O fatiamento [::-1] gera uma cópia invertida da lista original
    # A comparação lógica '==' checa se o padrão é simétrico (palíndromo)
    if fragmento == fragmento[::-1]:
        print(f"A sequência {fragmento} é um palíndromo!")
    else:
        print(f"A sequência {fragmento} NÃO é um palíndromo.")
else:
    print("Nenhuma sequência foi inserida.")
```
