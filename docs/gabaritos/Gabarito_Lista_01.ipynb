### - Exercício 1

### - Exercício 2
O comportamento dos operadores matemáticos muda dependendo do tipo de dado que estamos usando. Escreva um programa que peça ao usuário para digitar um número inteiro qualquer. O programa deve salvar essa entrada como um texto (String) e como um número inteiro (Int). Em seguida, o programa deve multiplicar ambas as variáveis por 3 e mostrar os dois resultados na tela.

```python
# RESOLUÇÃO - EXERCÍCIO 2

# Entrada de dados (O input sempre recebe como String inicialmente)
entrada_string = input("Digite um número: ")

# Conversão de tipo (Casting para Inteiro)
entrada_inteiro = int(entrada_string)

# Processamento e Saída (String repete o texto, Int faz a conta matemática)
print(f"Resultado String: {entrada_string * 3}")
print(f"Resultado Inteiro: {entrada_inteiro * 3}")
```

###- Exercício 3

Um aluno de Engenharia precisou criar um programa para calcular o custo final de um serviço. A regra era simples: pegar o custo do material, multiplicar por 2 e somar uma taxa de entrega informada pelo usuário. O código que ele escreveu foi:

custo_material = input("Digite o custo do material: ")

taxa_entrega = input("Digite a taxa extra de entrega: ")

custo_total = custo_material * 2 + taxa_entrega

print(f"O custo total do projeto é de R$ {custo_total}")

a) Explique por que o computador gerou um resultado diferente do esperado.

b) Reescreva o código do aluno corrigindo o problema.

```python
# RESOLUÇÃO - EXERCÍCIO 3

# Resposta a): O erro ocorreu porque a função input() lê tudo como Texto (String).
# Ao multiplicar a string "500" por 2, o Python repetiu o texto ("500500").
# Ao somar com "15", ele apenas concatenou (juntou) as palavras, gerando "50050015".

# Resposta b): Código corrigido (Convertendo para float)
custo_material = float(input("Digite o custo do material: "))
taxa_entrega = float(input("Digite a taxa extra de entrega: "))

custo_total = custo_material * 2 + taxa_entrega

print(f"O custo total do projeto é de R$ {custo_total:.2f}")
```

### - Exercício 4

Escreva um programa que leia a quantidade de dias, horas, minutos e segundos de um experimento. O programa deve calcular e mostrar o tempo total do experimento expresso apenas em segundos.

```python
# RESOLUÇÃO - EXERCÍCIO 4

dias = int(input("Dias: "))
horas = int(input("Horas: "))
minutos = int(input("Minutos: "))
segundos = int(input("Segundos: "))

# Cálculo: converter tudo para a base de segundos
total_segundos = (dias * 86400) + (horas * 3600) + (minutos * 60) + segundos

print(f"O tempo total do experimento foi de {total_segundos} segundos.")
```

### - Exercício 5

Na manipulação de textos, o fatiamento (slicing) nos permite extrair pedaços específicos de uma string utilizando a estrutura texto[inicio:fim]. Escreva um programa que peça ao usuário para digitar uma data de nascimento no formato DD/MM/AAAA (exatamente com 10 caracteres). O programa deve utilizar o fatiamento para extrair e imprimir separadamente o dia, o mês e o ano.

```python
# RESOLUÇÃO - EXERCÍCIO 5

data = input("Digite a data de nascimento (DD/MM/AAAA): ")

# Fatiamento baseado nos índices da string (inclui o primeiro e retira o ultimo)
dia = data[0:2]
mes = data[3:5]
ano = data[6:10]

print(f"Dia: {dia}")
print(f"Mês: {mes}")
print(f"Ano: {ano}")
```

### - Exercício 6

Um laboratório de química precisa armazenar frascos de um novo reagente em caixas. Escreva um programa que peça ao usuário dois valores: o número total de frascos produzidos e a capacidade da caixa (quantos frascos cabem em uma única caixa). Calcule e mostre na tela quantas caixas ficarão completamente cheias e quantos frascos sobrarão fora das caixas completas.

```python
# RESOLUÇÃO - EXERCÍCIO 6

total_frascos = int(input("Total de frascos: "))
capacidade = int(input("Capacidade da caixa: "))

# Divisão inteira para as caixas e resto para a sobra
caixas_cheias = total_frascos // capacidade
frascos_sobra = total_frascos % capacidade

print(f"Caixas completas: {caixas_cheias}")
print(f"Frascos que sobraram: {frascos_sobra}")
```

### - Exercício 7

Escreva um programa que funcione como uma calculadora de volumes para três formas geométricas: um cubo, um cilindro e uma pirâmide de base quadrada. O programa deve pedir ao usuário que digite três valores (em metros): uma medida de lado, uma de raio e uma de altura. Em seguida, ele deve calcular e exibir os volumes utilizando as seguintes fórmulas: Cubo: V = lado³; Cilindro: V = π × raio² × altura (Utilize 3.14 como valor de π); Pirâmide: V = (lado² × altura) / 3.

```python
# RESOLUÇÃO - EXERCÍCIO 7

# Lendo entradas
lado = float(input("Digite o valor do lado (m): "))
raio = float(input("Digite o valor do raio (m): "))
altura = float(input("Digite o valor da altura (m): "))

# Calculando o volume de cada uma das formas
vol_cubo = lado ** 3
vol_cilindro = 3.14 * (raio ** 2) * altura
vol_piramide = (lado ** 2 * altura) / 3

# Imprimindo os respectivos valores
print(f"O volume do cubo é: {vol_cubo:.2f} m³")
print(f"O volume do cilindro é: {vol_cilindro:.2f} m³")
print(f"O volume da pirâmide é: {vol_piramide:.2f} m³")
```

### - Exercício 8

Escreva um programa que receba um número inteiro de três dígitos. Utilizando lógica matemática, o programa deve "desmembrar" o número, extrair a centena, a dezena e a unidade, e no final imprimir a soma desses três algarismos. Dica: Utilize divisão inteira por 10 e o resto para encontrar os algarismos.

```python
# RESOLUÇÃO - EXERCÍCIO 8

numero = int(input("Digite um número de 3 dígitos: "))

# Extração matemática das posições
centena = numero // 100
dezena = (numero % 100) // 10
unidade = numero % 10

soma_total = centena + dezena + unidade

print(f"Centena: {centena}")
print(f"Dezena: {dezena}")
print(f"Unidade: {unidade}")
print(f"A soma dos dígitos é: {soma_total}")
```

### - Exercício 10

Um experimento em um laboratório automatizado, os pesquisadores precisam de um programa que calcule automaticamente a que horas um experimento vai terminar. Crie um programa que peça ao usuário três valores inteiros: a hora inicial (0 a 23), o minuto inicial (0 a 59) e a duração do experimento em minutos. O programa deve calcular e imprimir a hora e o minuto exatos em que o experimento terminará, respeitando o formato de um relógio de 24 horas.

```python
# RESOLUÇÃO - EXERCÍCIO 10

# Lendo entradas
hora_inicial = int(input("Hora inicial: "))
minuto_inicial = int(input("Minuto inicial: "))
duracao = int(input("Duração (em minutos): "))

# Soma total e cálculo dos ciclos
total_minutos = minuto_inicial + duracao
minuto_final = total_minutos % 60
horas_extras = total_minutos // 60

# Aplicação do módulo 24 para as horas
hora_final = (hora_inicial + horas_extras) % 24

print(f"Hora final: {hora_final}")
print(f"Minuto final: {minuto_final}")
```
